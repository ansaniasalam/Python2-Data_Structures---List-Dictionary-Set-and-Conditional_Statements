## List (Creation, Modification and Access):  

### 1. List Creation: 
##### a. Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28] 
##### b. Create a list named name_list with five string elements. 

```python
# a. Create a list named age_list with five integer elements.

age_list = [24, 25, 26, 27, 28]
print(age_list)
```

```python
# b. Create a list named name_list with five string elements.

name_list = ["Asha", "Ben", "Celine", "Dev", "Esha"]
print(name_list)
```

### 2. List Operations / Modifications: 
##### a. Append the string "Yazhini" to name_list. 
##### b. Insert the element 30 at index 2 in age_list.  
##### c. Remove the string "Yazhini" from name_list. 
##### d. Pop the last element from age_list.  
##### e. Extend the age_list with additional ages [29, 30, 26]. 
##### f. Sort age_list in descending order. 
##### g. Find Max age, Min age and sum of all ages from age_list.

```python
# a. Append the string "Yazhini" to name_list.

name_list = ["Asha", "Ben", "Celine", "Dev", "Esha"]
name_list.append("Yazhini")
print(name_list)
```

```python
# b. Insert the element 30 at index 2 in age_list.  

age_list = [24, 25, 26, 27, 28]
age_list.insert(2, 30)
print(age_list)
```

```python
# c. Remove the string "Yazhini" from name_list. 

name_list = ["Asha", "Ben", "Celine", "Dev", "Esha", "Yazhini"]
name_list.remove("Yazhini")
print(name_list)
```

```python
# d. Pop the last element from age_list.  

age_list = [24, 25, 30, 26, 27, 28]
age_list.pop()
print(age_list)
```

```python
# e. Extend the age_list with additional ages [29, 30, 26].

age_list = [24, 25, 30, 26, 27]
age_list.extend([29, 30, 26])
print(age_list)
```

```python
# f. Sort age_list in descending order.

age_list = [24, 25, 30, 26, 27, 29, 30, 26]
age_list.sort(reverse=True)
print(age_list)
```

```python
# g. Find Max age, Min age and sum of all ages from age_list.

age_list = [30, 30, 29, 27, 26, 26, 25, 24]
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of all ages:", sum(age_list))
```

## Dictionary (Creation, Modification and Access): 

##### a. Create a dictionary named student_marks that maps the names of five students to                their marks (use scale of from 0 to 100). 
##### b. Access and print the mark of a specific student, of your choice. 
##### c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary. 
##### d. Update the mark of any one older student to 82. 
##### e. Use the keys(), values(), and items() methods to print all keys, values, and key-              value pairs in the student_marks dictionary.

```python
# a. Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).

student_marks = {"Asha": 88, "Ben": 76,"Celine": 94,  "Dev": 81, "Esha": 69}
print(student_marks)
```

```python
# b. Access and print the mark of a specific student, of your choice.

student_marks = {"Asha": 88, "Ben": 76,"Celine": 94,"Dev": 81,"Esha": 69}
print(student_marks["Celine"])
```

```python
# c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.
 
student_marks = {"Asha": 88, "Ben": 76,"Celine": 94,"Dev": 81,"Esha": 69}
student_marks.update({"Janani": 80})
print(student_marks)

# or
student_marks = {"Asha": 88, "Ben": 76,"Celine": 94,"Dev": 81,"Esha": 69}
student_marks["Janani"] = 80
print(student_marks)
```

```python
# d. Update the mark of any one older student to 82.

student_marks = {"Asha": 88, "Ben": 76,"Celine": 94,"Dev": 81,"Esha": 69}
student_marks.update({"Ben": 82})
print(student_marks)

# or
student_marks = {"Asha": 88, "Ben": 76,"Celine": 94,"Dev": 81,"Esha": 69}
student_marks["Ben"] = 82
print(student_marks)
```

```python
# e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.

student_marks = {'Asha': 88, 'Ben': 82, 'Celine': 94, 'Dev': 81, 'Esha': 69}
print("Student names:", student_marks.keys())
print("Marks:", student_marks.values())
print("Student and mark pairs:", student_marks.items())
```

### Performance Category Program: 

##### 1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

##### 2. Find the performance category based on the input score using following criteria: 
##### -- a. Above Average:  Score greater than 7 
##### -- b. Average: Score between 4 and 7(both inclusive) 
##### -- c. Below Average: Score lesser than 4 

##### 3. Output: Print the Performance category 

##### 4. Additional Step: You can give a prompt of your choice to each category.  
##### -- For eg: If score below average “Need to Improve your performance, consistent practice will lead to better results”. 
##### -- Sample Output: 
###### -- Enter your score (0 to 10): 7 
###### -- Average: Good effort! Keep practicing, there's room for improvement.

```python
# 1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
score = int(input("Enter a score from 0 to 10: "))
```

```python
# 2. Find the performance category based on the input score using following criteria: 
  # a. Above Average:  Score greater than 7 
  # b. Average: Score between 4 and 7(both inclusive) 
  # c. Below Average: Score lesser than 4 

score = int(input("Enter a score from 0 to 10: "))
if score < 0 or score > 10:
    print("Invalid score. Enter a score from 0 to 10.")
elif score > 7:
    print("Above Average")
elif score >= 4:
    print("Average")
else:
    print("Below Average")
```

```python
# also can include the entered score in the output by using an f-string:

score = int(input("Enter a score from 0 to 10: "))
if score < 0 or score > 10:
    print(f"Score entered: {score} — invalid. Enter a score from 0 to 10.")
elif score > 7:
    print(f"Score entered: {score} — Above Average")
elif score >= 4:
    print(f"Score entered: {score} — Average")
else:
    print(f"Score entered: {score} — Below Average")
```

```python
# 3. Output: Print the Performance category 

score = int(input("Enter a score from 0 to 10: "))
if score > 7:
    category = "Above Average"
elif score >= 4:
    category = "Average"
else:
    category = "Below Average"

print("Performance category:", category)
```

```python
# 4. Additional Step: You can give a prompt of your choice to each category.  
# For eg: If score below average “Need to Improve your performance, consistent practice will lead to better results”.

score = int(input("Enter your score (0 to 10): "))
if score > 7:
    print("Above Average: Excellent work! Keep it up.")
elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("Below Average: Consistent practice will help improve your performance.")
```
